### 3. 常见“疑难杂症”速查

*   **Q: 数据量太少怎么办？**
    *   分类/回归：用 **SVM** 或 **Naive Bayes** (Part 3)。
    *   深度学习：用 **Transfer Learning (迁移学习)** 或 **Few-Shot Learning** (Part 4/5)。
    *   数据增强：用 **GAN** 生成数据 (Part 4) 或 **SMOTE** (Part 3)。

*   **Q: 特征维度太高（几千列）怎么办？**
    *   降维：**PCA** (Part 3)。
    *   特征选择：**Lasso Regression** (Part 3)。
    *   Embedding：**Word2Vec** (Part 5) 或 **AutoEncoder** (Part 9)。

*   **Q: 模型效果不好，怎么调参？**
    *   暴力法：**Grid Search**。
    *   聪明法：**Bayesian Optimization** 或 **TPE (Hyperopt)** (Part 10)。

*   **Q: 需要解释模型为什么这么预测（给老板看）？**
    *   首选：**Decision Tree** (Part 3) 或 **Linear Regression** (Part 3)。
    *   复杂模型解释：使用 **SHAP** 或 **LIME** (Part 3 补充)。

*   **Q: 只有正样本（只有点击，没有“不点击”）怎么办？**
    *   推荐系统：使用 **One-Class SVM** 或 **BPR (Bayesian Personalized Ranking)** (Part 9)。
    *   异常检测：**Isolation Forest** (Part 3)。

这两个领域构成了现代 AI 的“探索”能力。为了满足“50种”的需求，我将它们细分为 **5 大家族**，涵盖了从经典的数据挖掘算法到最前沿的深度强化学习策略。

### 第一族：聚类算法 (Clustering)
*核心任务：在没有标签的情况下，将相似的数据自动归类。*

| 序号 | 算法名称 | 特性 | 典型场景 | 核心原理 |
| :--- | :--- | :--- | :--- | :--- |
| 1 | **K-Means** | 经典、快速 | 用户分群、图像色彩压缩 | 迭代计算质心，将样本分给最近的质心。需预设 K 值。 |
| 2 | **Mini-Batch K-Means** | 超大规模 | 实时在线聚类 | K-Means 的变体，每次仅使用随机抽取的一小批数据更新质心。 |
| 3 | **K-Medoids (PAM)** | 抗噪 | 含离群点的数据 | 类似 K-Means，但质心必须是真实存在的样本点（Medoid），而非计算出的均值。 |
| 4 | **Mean Shift** | 自动定类 | 图像分割、目标跟踪 | 基于核密度估计，数据点向密度最高的区域（模态中心）漂移。无需预设 K。 |
| 5 | **DBSCAN** | 密度、非球形 | 地理位置聚类、异常剔除 | 基于密度的空间聚类，能发现任意形状的簇，自动识别噪音点。 |
| 6 | **OPTICS** | 变密度 | 复杂密度分布的数据 | DBSCAN 的改进版，解决了不同密度簇难以同时被发现的问题。 |
| 7 | **HDBSCAN** | 层次密度 | 现代最佳通用聚类 | 结合了层次聚类和 DBSCAN，稳定性极强（需独立库 `hdbscan`）。 |
| 8 | **Agglomerative Clustering** | 层次化 | 基因序列、分类树 | 自底向上，两两合并最近的簇，形成树状结构。 |
| 9 | **BIRCH** | 内存友好 | 极大数据集（无法一次载入内存） | 构建聚类特征树 (CF Tree)，只读一遍数据即可完成聚类。 |
| 10 | **Spectral Clustering** | 图论方法 | 社交网络圈子发现 | 基于拉普拉斯矩阵的特征分解，将聚类问题转化为图切分问题。 |
| 11 | **Affinity Propagation** | 消息传递 | 图像、基因表达分析 | 所有数据点相互“传递消息”来争夺代表权（Exemplar），无需预设 K。 |
| 12 | **Gaussian Mixture Model (GMM)** | 软聚类 | 语音识别、金融风控 | 假设数据由多个高斯分布混合生成，计算样本属于某簇的概率。 |

### 第二族：降维与流形学习 (Dimensionality Reduction)
*核心任务：在保留信息的前提下，将高维数据压缩到低维，用于可视化或去噪。*

| 序号 | 算法名称 | 特性 | 典型场景 | 核心原理 |
| :--- | :--- | :--- | :--- | :--- |
| 13 | **PCA (Principal Component Analysis)** | 线性、方差最大 | 数据预处理、压缩 | 正交变换，寻找数据方差最大的方向作为主成分。 |
| 14 | **Kernel PCA** | 非线性 | 复杂非线性数据降维 | 引入核函数（如 RBF），将数据映射到高维空间后再做 PCA。 |
| 15 | **Truncated SVD (LSA)** | 稀疏矩阵 | 文本挖掘 (TF-IDF 降维) | 奇异值分解的变体，专门处理稀疏矩阵。 |
| 16 | **NMF (Non-Negative Matrix Factorization)** | 可解释性强 | 话题模型、图像特征分解 | 强制矩阵分解后的分量非负，分解出的特征具有物理意义（如“眼睛”、“鼻子”）。 |
| 17 | **ICA (Independent Component Analysis)** | 盲源分离 | 鸡尾酒会问题（分离混音） | 寻找统计独立的成分，而非不相关的成分（常用于信号处理）。 |
| 18 | **Factor Analysis** | 隐变量 | 心理学问卷分析 | 假设观察变量由潜在的隐变量（因子）和噪音生成。 |
| 19 | **t-SNE** | 强可视化 | 高维数据（如 MNIST）可视化 | 保持高维空间中相似点的概率分布在低维空间中一致。 |
| 20 | **UMAP** | 快、保结构 | 细胞数据分析、通用可视化 | 比 t-SNE 更快，且能更好地保留数据的全局结构（需独立库 `umap-learn`）。 |
| 21 | **Isomap** | 等距映射 | 瑞士卷数据集展开 | 保持数据点之间的测地距离（Geodesic distance）。 |
| 22 | **LLE (Locally Linear Embedding)** | 局部线性 | 图像流形展开 | 保持每个点与其邻居的线性关系在低维空间不变。 |
| 23 | **MDS (Multidimensional Scaling)** | 距离保持 | 城市间距离还原地图 | 尽可能在低维空间保持样本间的欧氏距离。 |

### 第三族：异常检测 (Anomaly Detection)
*核心任务：在无标签数据中找出“格格不入”的样本。*

| 序号 | 算法名称 | 特性 | 典型场景 | 核心原理 |

| 26 | **Local Outlier Factor (LOF)** | 密度敏感 | 局部密度异常检测 | 比较样本点与其邻居的密度，若密度显著低于邻居则为异常。 |
| 27 | **Elliptic Envelope** | 鲁棒统计 | 符合高斯分布的数据 | 拟合一个鲁棒的协方差估计，假设正常数据服从高斯分布。 |

### 💻 Python 实现示例

#### 1. 无监督学习：DBSCAN 聚类
```python
from sklearn.cluster import DBSCAN
from sklearn.datasets import make_moons
import matplotlib.pyplot as plt

# 生成非球形数据（月亮形状）
X, _ = make_moons(n_samples=300, noise=0.05, random_state=42)

# 使用 DBSCAN (无需指定簇数量，指定半径 eps 和最小样本数)
# K-Means 在这种数据上会失效
db = DBSCAN(eps=0.3, min_samples=5).fit(X)
labels = db.labels_

# 绘图
plt.scatter(X[:, 0], X[:, 1], c=labels, cmap='plasma')
plt.title("DBSCAN Clustering")
plt.show()
```

#### 2. 强化学习：PPO (使用 Stable-Baselines3)
这是目前最流行的 RL 库，代码极其简洁。
```python
# pip install stable-baselines3 gym shimmy
import gymnasium as gym
from stable_baselines3 import PPO

# 1. 创建环境 (CartPole: 倒立摆平衡)
env = gym.make("CartPole-v1")

# 2. 实例化 Agent (使用 PPO 算法)
model = PPO("MlpPolicy", env, verbose=1)

# 3. 训练
print("开始训练...")
model.learn(total_timesteps=10000)

# 4. 测试
obs, _ = env.reset()
for i in range(100):
    action, _states = model.predict(obs, deterministic=True)
    obs, reward, terminated, truncated, info = env.step(action)
    if terminated or truncated:
        obs, _ = env.reset()
env.close()
```

### 第一族：基础组件与前馈网络 (Foundations & Feedforward)
*这是深度学习的基石，所有复杂的模型都是由这些组件搭建而成的。*

| 序号 | 算法/组件名称 | 特性 | 典型场景 | 核心原理 |
| :--- | :--- | :--- | :--- | :--- |

### 💻 Python 实现示例 (PyTorch)

这是深度学习的“Hello World”：定义一个简单的卷积神经网络 (CNN) 来识别图像。

```python
import torch
import torch.nn as nn
import torch.optim as optim
import torch.nn.functional as F

# 1. 定义网络结构 (一个简单的类似 LeNet 的 CNN)
class SimpleCNN(nn.Module):
    def __init__(self):
        super(SimpleCNN, self).__init__()
        # 卷积层: 输入1通道(黑白), 输出32通道, 核大小3x3
        self.conv1 = nn.Conv2d(1, 32, 3, 1)
        # 卷积层: 输入32通道, 输出64通道
        self.conv2 = nn.Conv2d(32, 64, 3, 1)
        # Dropout: 防止过拟合
        self.dropout1 = nn.Dropout(0.25)
        self.dropout2 = nn.Dropout(0.5)
        # 全连接层
        self.fc1 = nn.Linear(9216, 128)
        self.fc2 = nn.Linear(128, 10) # 10个数字分类

    def forward(self, x):
        # 前向传播逻辑
        x = self.conv1(x)
        x = F.relu(x)
        x = self.conv2(x)
        x = F.relu(x)
        x = F.max_pool2d(x, 2) # 最大池化
        x = self.dropout1(x)
        x = torch.flatten(x, 1) # 展平
        x = self.fc1(x)
        x = F.relu(x)
        x = self.dropout2(x)
        x = self.fc2(x)
        output = F.log_softmax(x, dim=1)
        return output

# 2. 实例化与测试
model = SimpleCNN()
print(model)

# 假设输入一张 28x28 的黑白图片 (Batch Size=1)
dummy_input = torch.randn(1, 1, 28, 28)
output = model(dummy_input)
print(f"输出形状: {output.shape}") # torch.Size([1, 10])
```

### 💻 Python 实现示例 (OpenCV)

这里展示一个经典的**特征匹配 (Feature Matching)** 任务。这是全景拼接、图像搜索和 AR 的核心技术。我们将使用 **ORB** 算法，因为它免费、快速且效果好。

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# 1. 读取两张图片 (假设你有两张不同角度拍摄的同一物体图片)
# 这里我们手动生成两个简单的图像来演示
img1 = np.zeros((300, 300), dtype='uint8')
cv2.rectangle(img1, (50, 50), (250, 250), 255, -1) # 画个矩形
cv2.circle(img1, (150, 150), 50, 0, -1)            # 挖个洞

# img2 是 img1 旋转并缩放后的版本
M = cv2.getRotationMatrix2D((150, 150), 45, 0.8)   # 旋转45度，缩放0.8
img2 = cv2.warpAffine(img1, M, (300, 300))

# 2. 初始化 ORB 检测器
orb = cv2.ORB_create()

# 3. 寻找关键点 (Keypoints) 和 描述符 (Descriptors)
kp1, des1 = orb.detectAndCompute(img1, None)
kp2, des2 = orb.detectAndCompute(img2, None)

# 4. 创建匹配器 (Brute-Force Matcher)
# ORB 使用二进制描述符，所以用 Hamming 距离
bf = cv2.BFMatcher(cv2.NORM_HAMMING, crossCheck=True)

# 5. 执行匹配
matches = bf.match(des1, des2)

# 6. 按照距离排序 (越小越好)
matches = sorted(matches, key=lambda x: x.distance)

# 7. 绘制前 20 个匹配结果
img_matches = cv2.drawMatches(img1, kp1, img2, kp2, matches[:20], None, 
                              flags=cv2.DrawMatchesFlags_NOT_DRAW_SINGLE_POINTS)

# 8. 显示结果
plt.figure(figsize=(12, 6))
plt.imshow(img_matches)
plt.title(f"ORB Feature Matching (Top {len(matches)} matches)")
plt.axis('off')
plt.show()
```

### 👁️ 视觉算法选型指南

*   **简单的颜色提取/形状分析**：不要用深度学习！用 **HSV 颜色空间转换 + 阈值分割 + 轮廓查找**。OpenCV 几行代码就能搞定，且能在树莓派上跑 60FPS。
*   **文档扫描/二维码**：用 **Adaptive Thresholding** 或 **Otsu**，配合 **Perspective Transform** 做矫正。
*   **人脸检测**：
    *   极低算力设备：**Haar Cascades**。
    *   一般设备：**MTCNN** 或 **Ultra-Light-Fast-Generic-Face-Detector-1MB** (深度学习)。
*   **图像拼接**：**SIFT** (精度高) 或 **ORB** (速度快) + **Homography**。
*   **去噪**：**Bilateral Filter** 是最好的非深度学习选择，因为它能保留边缘。

### 💻 Python 实现示例

这里展示两个极端的例子：
1.  **古典派**：使用 `sklearn` 进行 TF-IDF + 逻辑回归分类。
2.  **现代派**：使用 `transformers` 调用预训练 BERT。

#### 1. 古典 NLP：TF-IDF 文本分类
```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.pipeline import make_pipeline

# 1. 准备数据
texts = [
    "The dog is playing in the park",    # 宠物
    "Cat loves to sleep on the sofa",    # 宠物
    "Stock market is crashing today",    # 金融
    "Investment in gold is profitable"   # 金融
]
labels = ['pet', 'pet', 'finance', 'finance']

# 2. 构建管道：TF-IDF -> 贝叶斯分类器
model = make_pipeline(TfidfVectorizer(stop_words='english'), MultinomialNB())

# 3. 训练
model.fit(texts, labels)

# 4. 预测
print(model.predict(["I want to buy some stocks"])) # 输出: ['finance']
```

#### 2. 现代 NLP：使用 BERT 提取特征 (HuggingFace)
```python
# pip install transformers torch
from transformers import BertTokenizer, BertModel
import torch

# 1. 加载预训练模型和分词器
model_name = 'bert-base-uncased'
tokenizer = BertTokenizer.from_pretrained(model_name)
model = BertModel.from_pretrained(model_name)

# 2. 准备文本
text = "Artificial Intelligence is amazing."
inputs = tokenizer(text, return_tensors="pt")

# 3. 前向传播
with torch.no_grad():
    outputs = model(**inputs)

# 4. 获取句向量 (CLS token 的输出)
# last_hidden_state shape: [1, 6, 768] (Batch, Sequence Length, Hidden Size)
cls_embedding = outputs.last_hidden_state[:, 0, :] 

print(f"文本向量维度: {cls_embedding.shape}") # torch.Size([1, 768])
```

*   **需要理解语义/同义词**：

### 第一族：遍历与搜索 (Traversal & Search)

| 序号 | 算法名称 | 特性 | 典型场景 | 核心原理 |
| :--- | :--- | :--- | :--- | :--- |
| 1 | **BFS (Breadth-First Search)** | 广度优先 | 寻找无权图最短路径、社交好友推荐 | 层层递进，像水波纹一样扩散。使用队列 (Queue)。 |
| 2 | **DFS (Depth-First Search)** | 深度优先 | 迷宫寻路、拓扑排序 | 一条路走到黑，撞墙再回头。使用栈 (Stack) 或递归。 |
| 3 | **Bidirectional Search** | 双向搜索 | 两人相遇问题 | 同时从起点和终点开始搜索，在中间相遇，大幅减少搜索空间。 |

| 5 | **Best-First Search** | 贪心搜索 | 简单寻路 | 总是选择当前看起来离目标最近的节点（不保证最优解）。 |
| 6 | **Beam Search** | 束搜索 | 机器翻译、生成模型 | 限制宽度的 BFS，每层只保留前 K 个最优节点，牺牲完备性换取速度。 |

| 9 | **Bellman-Ford** | 含负权边 | 金融套利检测 | 对所有边进行 $V-1$ 次松弛操作，能检测负权环。 |

### 💻 Python 实现示例 (NetworkX)

`networkx` 是 Python 图论领域的绝对霸主，虽然它在大图上速度不如 `igraph` 或 `graph-tool`，但其 API 设计极其人性化，适合学习和原型开发。

#### 示例：寻找《权力的游戏》人物关系中的关键人物

```python
import networkx as nx
import matplotlib.pyplot as plt

# 1. 创建一个图 (模拟人物关系)
G = nx.Graph()
relationships = [
    ('Jon', 'Sansa'), ('Jon', 'Arya'), ('Jon', 'Tyrion'), ('Jon', 'Daenerys'),
    ('Sansa', 'Arya'), ('Sansa', 'Tyrion'), ('Sansa', 'Littlefinger'),
    ('Tyrion', 'Daenerys'), ('Tyrion', 'Varys'), ('Daenerys', 'Jorah'),
    ('Arya', 'Gendry'), ('Jaime', 'Cersei'), ('Cersei', 'Tyrion')
]
G.add_edges_from(relationships)

# 2. 基础分析
print(f"节点数: {G.number_of_nodes()}")
print(f"边数: {G.number_of_edges()}")

# 3. 寻找最短路径 (Jon 到 Cersei)
path = nx.shortest_path(G, source='Jon', target='Cersei')
print(f"Jon 到 Cersei 的最短路径: {path}") 
# 输出: ['Jon', 'Tyrion', 'Cersei']

# 4. 计算中心性 (谁最重要？)
# PageRank
pr = nx.pagerank(G)
# Betweenness (桥梁作用)
bc = nx.betweenness_centrality(G)

print("\n--- 关键人物 (PageRank) ---")
for name, score in sorted(pr.items(), key=lambda x: x[1], reverse=True)[:3]:
    print(f"{name}: {score:.4f}")

# 5. 社区发现 (使用 Louvain 算法)
# NetworkX 2.x+ 内置了 community 模块，或者使用 python-louvain 库
try:
    from networkx.algorithms.community import louvain_communities
    communities = louvain_communities(G)
    print(f"\n--- 发现 {len(communities)} 个社区 ---")
    for i, comm in enumerate(communities):
        print(f"社区 {i+1}: {comm}")
except ImportError:
    print("需要 networkx 2.8+ 支持 louvain_communities")

# 6. 可视化
plt.figure(figsize=(10, 8))
pos = nx.spring_layout(G, seed=42) # 弹簧布局
nx.draw(G, pos, with_labels=True, node_color='lightblue', 
        node_size=2000, font_size=12, edge_color='gray')
plt.title("Game of Thrones Interaction Graph")
plt.show()
```

### 🕸️ 图算法选型指南

*   **我要找两点间怎么走最近**：
    *   地图/导航：**Dijkstra** (无负权) 或 **A*** (有坐标启发)。
    *   有负权边（如套利）：**Bellman-Ford**。
    *   所有点对所有点：**Floyd-Warshall** (节点数 < 500)。
*   **我要分析谁最重要**：
    *   明星人物（连接多）：**Degree Centrality**。
    *   八卦中心（传得快）：**Closeness Centrality**。
    *   控制关卡（必经之路）：**Betweenness Centrality**。
    *   权威认证（谷歌排名）：**PageRank**。
*   **我要划分圈子**：
    *   不知道分几类：**Louvain** 或 **Leiden** (目前最强)。
    *   只想要简单的切分：**Spectral Clustering**。
*   **我要处理超大规模图 (千万节点)**：
    *   不要用 `networkx`（纯 Python 太慢）。
    *   使用 **`igraph`** (C++底层)、**`graph-tool`** 或 **`Snap.py`**。

### 第一族：平滑与基线模型 (Smoothing & Baselines)
*最简单的预测往往最有效，用于去除噪声或作为对照组。*

| 序号 | 算法名称 | 特性 | 典型场景 | 核心原理 |
| :--- | :--- | :--- | :--- | :--- |
| 1 | **SMA (Simple Moving Average)** | 简单移动平均 | 股票均线、趋势观察 | 计算过去 N 个时间点的算术平均值。 |
| 2 | **WMA (Weighted Moving Average)** | 加权移动平均 | 近期数据更重要的场景 | 给过去的数据分配不同的权重，越近权重越大。 |
| 3 | **EMA (Exponential Moving Average)** | 指数移动平均 | 金融技术指标 (MACD) | 权重随时间呈指数衰减，对最新变化反应更灵敏。 |
| 4 | **SES (Simple Exponential Smoothing)** | 单指数平滑 | 无趋势、无季节性的数据 | 预测值是上一时刻实际值和预测值的加权和。 |
| 5 | **Holt's Linear Trend** | 双指数平滑 | 有线性趋势的数据 | 在 SES 基础上增加了一个“趋势”分量平滑。 |
| 6 | **Holt-Winters** | 三指数平滑 | 有趋势且有季节性的数据 | 同时对水平、趋势和季节性 (Seasonality) 进行平滑。 |
| 7 | **Naive Drift** | 漂移法 | 随机游走基准 | 假设未来的变化趋势与过去的历史平均趋势一致。 |
| 8 | **Croston's Method** | 稀疏数据 | 备件库存预测 | 专门处理间歇性需求（很多 0 值）的时间序列。 |

### 💻 Python 实现示例

这里展示两个最常用的场景：**ARIMA 统计预测** 和 **Prophet 商业预测**。

#### 1. 使用 `statsmodels` 进行 SARIMA 预测
这是最“正统”的统计学方法。

```python
import pandas as pd
import matplotlib.pyplot as plt
from statsmodels.tsa.statespace.sarimax import SARIMAX

# 1. 创建模拟数据 (带季节性的趋势数据)
dates = pd.date_range(start='2020-01-01', periods=100, freq='M')
values = [x + (x%12)*2 + (x//12) for x in range(100)] # 线性趋势 + 周期
data = pd.Series(values, index=dates)

# 2. 定义并训练模型
# order=(p,d,q), seasonal_order=(P,D,Q,s)
# 这里假设周期为 12 (月度数据)
model = SARIMAX(data, order=(1, 1, 1), seasonal_order=(1, 1, 1, 12))
results = model.fit(disp=False)

# 3. 预测未来 12 个月
forecast = results.get_forecast(steps=12)
mean_forecast = forecast.predicted_mean
conf_int = forecast.conf_int()

# 4. 绘图
plt.figure(figsize=(12, 6))
plt.plot(data.index, data, label='History')
plt.plot(mean_forecast.index, mean_forecast, label='Forecast', color='red')
plt.fill_between(mean_forecast.index, conf_int.iloc[:, 0], conf_int.iloc[:, 1], color='pink', alpha=0.3)
plt.title("SARIMA Forecast")
plt.legend()
plt.show()
```

#### 2. 使用 `prophet` 进行傻瓜式预测
适合有节假日、突变和缺失值的商业数据。

```python
# pip install prophet
from prophet import Prophet
import pandas as pd

# Prophet 强制要求列名为 'ds' (时间) 和 'y' (值)
df = pd.DataFrame({'ds': dates, 'y': values})

# 1. 初始化模型 (可自动拟合周、年季节性)
m = Prophet(yearly_seasonality=True, daily_seasonality=False)

# 2. 添加节假日 (可选)
m.add_country_holidays(country_name='US')

# 3. 训练
m.fit(df)

# 4. 构建未来 dataframe
future = m.make_future_dataframe(periods=12, freq='M')

# 5. 预测
forecast = m.predict(future)

# 6. 绘图
fig1 = m.plot(forecast)
```

这是“Python 算法全景图谱”的**第八部分**。

这一部分聚焦于 **强化学习 (Reinforcement Learning, RL)**。

这是通往“通用人工智能 (AGI)”最令人兴奋的路径之一。不同于监督学习（有老师教）和无监督学习（自学），强化学习是**“从试错中学习”**。智能体 (Agent) 在环境 (Environment) 中采取行动 (Action)，获得奖励 (Reward) 或惩罚，从而不断优化自己的策略 (Policy)。

从 AlphaGo 击败人类棋手，到 ChatGPT 的 RLHF 对齐训练，RL 正在重塑 AI 的上限。

以下是 **50 种 Python 可实现的强化学习算法与核心概念**详解：

---

### 第一族：基础组件与表格型方法 (Tabular Methods)
*环境很小，可以用一张 Excel 表记下所有情况的“作弊小抄”。*

| 序号 | 算法/概念名称 | 类型 | 典型场景 | 核心原理 |
| :--- | :--- | :--- | :--- | :--- |
| 1 | **Markov Decision Process (MDP)** | 理论基础 | 所有 RL 问题 | 五元组 $(S, A, P, R, \gamma)$：状态、动作、转移概率、奖励、折扣因子。 |
| 2 | **Bellman Equation** | 核心方程 | 动态规划 | 当前状态的价值 = 即时奖励 + 折扣后的未来价值。 |
| 3 | **Dynamic Programming (DP)** | 规划 | 已知环境模型的迷宫 | 包含策略评估 (Policy Eval) 和策略提升 (Policy Improv)。 |
| 4 | **Monte Carlo (MC)** | 无模型 | 21点扑克游戏 | 玩完一整局游戏后，根据最终结果倒推每一步的价值（取平均）。 |
| 5 | **Temporal Difference (TD)** | 无模型 | 边走边学 | 结合了 DP 和 MC，每走一步就利用下一步的估计值更新当前步的价值。 |
| 6 | **Q-Learning** | Off-policy | 迷宫寻路 (GridWorld) | 学习 $Q(s,a)$ 表。更新时假设下一步会采取最优动作（贪婪），即使实际没这么做。 |
| 7 | **SARSA** | On-policy | 比较保守的寻路 | 学习 $Q(s,a)$ 表。更新时严格使用下一步实际采取的动作。 |
| 8 | **Epsilon-Greedy** | 探索策略 | 解决探索与利用困境 | 以 $\epsilon$ 的概率随机乱走（探索），以 $1-\epsilon$ 的概率走目前认为最好的路（利用）。 |

---

### 第二族：深度强化学习 - 价值近似 (Deep Q-Networks)
*状态太多表存不下？用神经网络来估算 Q 值。*

| 序号 | 算法名称 | 改进点 | 典型场景 | 核心原理 |
| :--- | :--- | :--- | :--- | :--- |
| 9 | **DQN (Deep Q-Network)** | 深度学习 | Atari 游戏 (打砖块) | 用 CNN 拟合 Q 函数。引入 **Replay Buffer (经验回放)** 和 **Target Network** 打破数据相关性。 |
| 10 | **Double DQN** | 修正偏差 | 解决 DQN 高估问题 | 动作选择和价值评估使用两个不同的网络，减少 Q 值过高估计。 |
| 11 | **Dueling DQN** | 结构优化 | 复杂动作环境 | 将网络拆分为 Value（状态价值）和 Advantage（动作优势）两路，最后合并。 |
| 12 | **Prioritized Experience Replay** | 采样优化 | 稀疏奖励任务 | 不再随机抽取经验，而是优先抽取那些“让模型惊讶”（TD Error 大）的样本。 |
| 13 | **Noisy Nets** | 探索优化 | 替代 Epsilon-Greedy | 在网络参数中加入噪声，实现更连贯、更系统的探索。 |
| 14 | **Rainbow** | 集大成者 | 各种游戏 SOTA | 融合了 Double, Dueling, PER, Noisy Nets, Distributional RL 等 7 种改进技术的终极版 DQN。 |
| 15 | **DRQN (Recurrent)** | 记忆能力 | 毁灭战士 (Doom) | 引入 LSTM 处理部分可观测环境 (POMDP)，让 Agent 有记忆。 |

---

### 第三族：策略梯度方法 (Policy Gradient)
*不猜分数了，直接学习“什么状态下该做什么动作”的概率分布。*

| 序号 | 算法名称 | 特性 | 典型场景 | 核心原理 |
| :--- | :--- | :--- | :--- | :--- |
| 16 | **REINFORCE** | 基础 PG | 简单的 CartPole | 蒙特卡洛策略梯度。玩完一局，如果赢了，就提高这一局所有动作的概率。 |
| 17 | **Actor-Critic (AC)** | 架构 | 绝大多数现代算法基石 | **Actor** (演员) 负责做动作，**Critic** (评论家) 负责打分。Actor 根据 Critic 的打分修改策略。 |
| 18 | **A2C (Advantage Actor-Critic)** | 同步并行 | 机器人控制 | 使用优势函数 $A(s,a) = Q(s,a) - V(s)$ 替代原始回报，降低方差。 |
| 19 | **A3C (Asynchronous A2C)** | 异步并行 | 多核 CPU 加速训练 | 多个 Agent 在不同线程并行交互，异步更新全局网络。 |
| 20 | **TRPO (Trust Region)** | 稳定性 | 复杂的连续控制 | 限制每次策略更新的幅度（KL 散度），防止步子迈太大导致策略崩溃。 |
| 21 | **PPO (Proximal Policy Opt)** | 工业标准 | OpenAI Five, ChatGPT | TRPO 的简化版。通过截断 (Clip) 目标函数来限制更新幅度，简单且效果极好。 |
| 22 | **DPPO** | 分布式 | 大规模集群训练 | 分布式版的 PPO。 |

### 第四族：确定性策略与连续控制 (Deterministic & Continuous)
*专门解决“机械臂关节角度”这种连续输出问题。*

| 序号 | 算法名称 | 特性 | 典型场景 | 核心原理 |
| :--- | :--- | :--- | :--- | :--- |
| 23 | **Dpg (Deterministic PG)** | 理论 | 连续动作空间 | 直接输出确定的动作值，而不是概率分布。 |
| 24 | **DDPG** | Deep版 DPG | 机械臂抓取 | DQN + Actor-Critic。Actor 输出动作，Critic 评估 Q 值。 |
| 25 | **TD3** | 改进版 DDPG | 稳定的机器人行走 | Twin Delayed DDPG。使用两个 Critic 取最小值，延迟更新 Actor，解决 DDPG 不稳定的问题。 |
| 26 | **SAC (Soft Actor-Critic)** | 最大熵 | 真实机器人训练 | 在最大化奖励的同时，最大化策略的熵（随机性），鼓励探索，鲁棒性极强。 |

### 第五族：基于模型的强化学习 (Model-Based RL)
*先在大脑里模拟世界怎么运作，再做决定。*

| 序号 | 算法名称 | 特性 | 典型场景 | 核心原理 |
| :--- | :--- | :--- | :--- | :--- |
| 27 | **Dyna-Q** | 混合 | 简单迷宫加速 | 既从真实经验中学习，也从“想象”（模型模拟）的经验中学习。 |
| 28 | **MCTS (Monte Carlo Tree Search)** | 搜索 | AlphaGo / 棋类 | 模拟未来的多种可能性，构建搜索树，选择胜率最高的分支。 |
| 29 | **World Models** | 幻觉训练 | 赛车游戏 | 训练一个 VAE+RNN 来模拟环境（做梦），Agent 在梦中训练，然后应用到现实。 |
| 30 | **MuZero** | 终极棋手 | 围棋、Atari、Shogi | AlphaZero 的进化版。不需要知道游戏规则，直接学习环境的隐含动态模型。 |
| 31 | **MBPO** | 样本效率 | 极少数据训练 | 在学到的模型上进行短周期的推演，生成大量虚拟数据用于训练 SAC。 |

### 第六族：多智能体强化学习 (Multi-Agent RL - MARL)
*一群 AI 在一起打群架或合作。*

| 序号 | 算法名称 | 模式 | 典型场景 | 核心原理 |
| :--- | :--- | :--- | :--- | :--- |

| 36 | **MAPPO** | 混合 | 多智能体协作 SOTA | PPO 的多智能体版本，同样采用 CTDE 架构。 |

### 💻 Python 实现示例 (OpenAI Gym + Stable Baselines3)

不要从头手写 PPO（除非为了学习），工业界和学术界通常使用成熟的库，如 `Stable Baselines3`。

#### 示例：训练 AI 玩 "CartPole" (倒立摆)

```python
# 需要安装: pip install stable-baselines3 gymnasium shimmy
import gymnasium as gym
from stable_baselines3 import PPO
from stable_baselines3.common.env_util import make_vec_env

# 1. 创建环境
# CartPole-v1: 任务是控制小车左右移动，保持杆子不倒
env = make_vec_env("CartPole-v1", n_envs=4)

# 2. 初始化模型 (PPO)
# MlpPolicy: 使用多层感知机 (MLP) 处理输入 (因为输入是坐标数值，不是图像)
# verbose=1: 打印训练进度
model = PPO("MlpPolicy", env, verbose=1)

# 3. 开始训练
print("开始训练...")
model.learn(total_timesteps=25000)
print("训练结束！")

# 4. 保存模型
model.save("ppo_cartpole")

# 5. 测试与可视化
# 注意：在 Jupyter/Colab 中渲染动画比较麻烦，通常需要额外配置
del model # 删除当前模型，演示加载
model = PPO.load("ppo_cartpole")

obs = env.reset()
print("开始测试推断...")
for _ in range(10):
    # predict 返回动作和状态(对于RNN)
    action, _states = model.predict(obs)
    # step 执行动作
    obs, rewards, dones, info = env.step(action)
    # env.render() # 如果在本地运行，取消注释可以看到动画窗口
    print(f"Step Reward: {rewards[0]}")
```

*   **连续动作控制（如机器人关节角度、方向盘力度）**：

    *   **CQL** (Offline RL)。
*   **需要对齐人类价值观 (LLM 微调)**：
    *   **RLHF (PPO)** 或 **DPO (Direct Preference Optimization)**。

这一部分聚焦于 **推荐系统 (Recommender Systems)**。

推荐系统是互联网产品的“印钞机”。从淘宝的“猜你喜欢”，到抖音的“刷不停”，再到 Netflix 的电影推荐，推荐算法决定了用户看到什么。它本质上是一个**信息过滤系统**，旨在解决信息过载问题。

以下是 **50 种 Python 可实现的推荐算法与核心技术**详解，涵盖了从经典的协同过滤到最前沿的深度推荐模型：

---

### 第一族：基于内容的推荐 (Content-Based Filtering)
*“因为你喜欢《黑客帝国1》，所以你应该也喜欢《黑客帝国2》。”*

| 序号 | 算法名称 | 特性 | 典型场景 | 核心原理 |
| :--- | :--- | :--- | :--- | :--- |
| 1 | **TF-IDF Similarity** | 文本匹配 | 新闻推荐 | 计算物品描述文本的 TF-IDF 向量，利用余弦相似度找相似物品。 |
| 2 | **Item Profile** | 标签匹配 | 电影推荐 | 为物品打上标签（科幻、动作），计算标签向量的重合度。 |
| 3 | **User Profile** | 用户画像 | 冷启动 | 根据用户注册信息（年龄、性别、兴趣）推荐符合画像的热门内容。 |
| 4 | **Knowledge Graph (KG)** | 知识关联 | 深度内容挖掘 | 利用知识图谱（如：导演->电影->演员）发现潜在的关联推荐。 |

---

### 第二族：协同过滤 (Collaborative Filtering - CF)
*“和你口味相似的人也喜欢这个。”*

| 序号 | 算法名称 | 类型 | 典型场景 | 核心原理 |
| :--- | :--- | :--- | :--- | :--- |
| 5 | **User-Based CF** | 记忆型 | 社交推荐 | 寻找相似用户，推荐他们喜欢但你没看过的物品。 |
| 6 | **Item-Based CF** | 记忆型 | 电商 (Amazon) | 寻找相似物品（被同一群人购买），推荐给你买过的物品的相似品。 |
| 7 | **Slope One** | 简化版 | 评分预测 | 基于物品之间平均评分差异的简单线性预测算法。 |
| 8 | **Jaccard Similarity** | 相似度 | 隐式反馈 | 计算两个集合（如购买记录）的交并比，用于计算相似度。 |
| 9 | **Pearson Correlation** | 相似度 | 显式评分 | 计算评分向量的相关系数，去除用户打分偏置（有的用户手松，有的手紧）。 |

### 第三族：矩阵分解 (Matrix Factorization)
*把巨大的评分矩阵拆解成“用户-隐向量”和“物品-隐向量”。*

| 序号 | 算法名称 | 优化目标 | 典型场景 | 核心原理 |
| :--- | :--- | :--- | :--- | :--- |
| 10 | **SVD (Singular Value Decomposition)** | 评分预测 | Netflix Prize | 将评分矩阵分解，提取潜在特征 (Latent Factors)。 |
| 11 | **SVD++** | 隐式信息 | 精度提升 | 在 SVD 基础上加入了“用户浏览过什么”这一隐式反馈信息。 |

### 💻 Python 实现示例 (Surprise & LightFM)

推荐系统通常分为 **召回 (Recall)** 和 **排序 (Ranking)** 两个阶段。

#### 1. 传统协同过滤 (使用 `scikit-surprise`)
适合评分预测任务。

```python
# pip install scikit-surprise
from surprise import SVD, Dataset, Reader
from surprise.model_selection import cross_validate

# 1. 准备数据 (用户ID, 物品ID, 评分)
# 这里使用内置的 MovieLens-100k 数据集
data = Dataset.load_builtin('ml-100k')

# 2. 选择算法 (SVD 矩阵分解)
algo = SVD()

# 3. 交叉验证评估
print("正在进行 5 折交叉验证...")
cross_validate(algo, data, measures=['RMSE', 'MAE'], cv=5, verbose=True)

# 4. 预测
trainset = data.build_full_trainset()
algo.fit(trainset)

uid = str(196)  # 用户 ID
iid = str(302)  # 电影 ID
pred = algo.predict(uid, iid, r_ui=4, verbose=True)
# 输出: user: 196 item: 302 r_ui = 4.00   est = 4.12   {'was_impossible': False}
```

#### 2. 混合推荐模型 (使用 `LightFM`)
同时利用协同过滤和内容特征（如标签），适合冷启动。

```python
# pip install lightfm
import numpy as np
from lightfm import LightFM
from lightfm.datasets import fetch_movielens

# 1. 获取数据 (包含物品特征)
data = fetch_movielens(min_rating=4.0)

# 2. 建立模型 (使用 WARP 损失函数优化排序)
model = LightFM(loss='warp')

# 3. 训练
# data['train'] 是稀疏交互矩阵
# data['item_features'] 是物品特征矩阵
model.fit(data['train'], item_features=data['item_features'], epochs=30, num_threads=2)

# 4. 推荐
def sample_recommendation(model, data, user_ids):
    n_users, n_items = data['train'].shape
    
    for user_id in user_ids:
        # 预测该用户对所有物品的分数
        scores = model.predict(user_id, np.arange(n_items), item_features=data['item_features'])
        
        # 取 Top 3
        top_items = data['item_labels'][np.argsort(-scores)]
        
        print(f"User {user_id}:")
        print(f"  推荐: {top_items[:3]}")

sample_recommendation(model, data, [3, 25, 450])
```

### 🎯 推荐系统选型指南

*   **刚起步，只有数据库里的交易记录**：
    *   **Item-Based CF** (如果物品少) 或 **Association Rules (Apriori)**。
    *   简单的统计推荐：“热销榜单”。
*   **有用户评分数据，追求精度**：
    *   **Matrix Factorization (SVD/ALS)**。
*   **只有点击/浏览数据 (隐式反馈)**：
    *   **ALS (Implicit)** 或 **BPR**。
*   **数据量大，特征丰富 (用户画像、上下文)**：
    *   **DeepFM** 或 **Wide & Deep**。这是目前大厂的主流排序模型。
*   **关注用户近期的兴趣变化**：
    *   **DIN** (阿里) 或 **SASRec** (序列推荐)。
*   **图结构数据 (社交网络/知识图谱)**：
    *   **LightGCN**。

这是“Python 算法全景图谱”的**第十部分**，也是**收官之作**。

这一部分聚焦于 **运筹优化与元启发式算法 (Operations Research & Meta-heuristics)**。

如果说机器学习是“预测未来”，那么运筹优化就是“**决策现在**”。如何在资源有限的情况下利润最大化？如何规划物流路径最短？如何调节 100 个参数让模型性能最好？

这部分涵盖了从严格的数学规划（求绝对最优解）到模拟自然的智能优化算法（求满意解）。

以下是 **50 种 Python 可实现的优化算法**详解：

### 第一族：数学规划 (Mathematical Programming)
*严谨的数学推导，寻找理论上的“绝对最优解”。*

| 序号 | 算法名称 | 类型 | 典型场景 | 核心原理 |
| :--- | :--- | :--- | :--- | :--- |
| 1 | **Linear Programming (LP)** | 线性规划 | 生产排程、饲料配比 | 目标函数和约束条件都是线性的。单纯形法 (Simplex) 是经典求解器。 |
| 2 | **MILP (Mixed-Integer LP)** | 混合整数 | 选址问题、排班 | 变量中部分必须是整数（如建厂数量不能是 1.5 个）。 |
| 3 | **Branch and Bound** | 分支定界 | 整数规划求解 | 将问题空间分解为子问题树，计算上下界，剪掉不可能包含最优解的分支。 |
| 4 | **Branch and Cut** | 分支切割 | 复杂的 TSP 问题 | 结合了分支定界和割平面法，通过添加有效不等式收紧松弛解。 |
| 5 | **Quadratic Programming (QP)** | 二次规划 | 投资组合优化 (Markowitz) | 目标函数是二次的（如最小化方差），约束是线性的。SVM 的对偶问题也是 QP。 |
| 6 | **Interior Point Method** | 内点法 | 大规模 LP/QP | 沿着可行域内部路径逼近最优解，比单纯形法在某些大规模问题上更快。 |
| 7 | **Dynamic Programming (DP)** | 动态规划 | 背包问题、最短路 | 将复杂问题分解为重叠子问题，存储中间状态避免重复计算。 |

*深度学习的引擎，沿着“山坡”往下滚。*

| 序号 | 算法名称 | 特性 | 典型场景 | 核心原理 |
| :--- | :--- | :--- | :--- | :--- |

| 9 | **SGD (Stochastic Gradient Descent)** | 一阶优化 | 神经网络训练 | 每次随机抽取一个样本计算梯度更新参数。 |
| 10 | **Momentum** | 动量加速 | 穿越平坦区域 | 引入物理“惯性”，抑制震荡，加速收敛。 |

| 14 | **L-BFGS** | 拟牛顿法 | 逻辑回归、风格迁移 | 近似计算海森矩阵的逆，内存占用少，适合中小规模的精确优化。 |

### 💻 Python 实现示例

这里展示两种截然不同的优化思路：
1.  **精确解**：使用 `scipy.optimize` 求解线性规划 (Linear Programming)。
2.  **近似解**：使用 `scipy.optimize` 的 **差分进化 (Differential Evolution)** 求解复杂的非凸函数。

#### 1. 线性规划 (生产利润最大化)
*场景：生产 A 和 B 两种产品，A 利润 3 元，B 利润 2 元。原料有限，求最大利润。*

```python
from scipy.optimize import linprog

# 目标：Max z = 3x + 2y
# scipy 默认是求 Min，所以系数取反：Min w = -3x - 2y
c = [-3, -2]

# 约束条件 (不等式 Ax <= b)：
# 2x + y <= 100 (原料1)
# x + y <= 80   (原料2)
A = [[2, 1], [1, 1]]
b = [100, 80]

# 变量边界 (x >= 0, y >= 0)
x_bounds = (0, None)
y_bounds = (0, None)

# 求解
res = linprog(c, A_ub=A, b_ub=b, bounds=[x_bounds, y_bounds], method='highs')

print("--- 线性规划结果 ---")
print(f"最优生产数量: 产品A = {res.x[0]:.1f}, 产品B = {res.x[1]:.1f}")
print(f"最大利润: {-res.fun:.1f}") # 结果取反回正
```

#### 2. 差分进化 (寻找复杂函数的全局最小值)
*场景：Rastrigin 函数（有很多局部最小值的波浪形函数），梯度下降法会卡死，需要进化算法。*

```python
import numpy as np
from scipy.optimize import differential_evolution
import matplotlib.pyplot as plt

# 定义 Rastrigin 函数 (典型的多峰函数)
def rastrigin(x):
    # x 是一个向量
    A = 10
    return A * len(x) + sum(x**2 - A * np.cos(2 * np.pi * x))

# 设定搜索范围 (-5.12 到 5.12)
bounds = [(-5.12, 5.12), (-5.12, 5.12)]

# 求解
# strategy='best1bin': 经典的 DE 策略
result = differential_evolution(rastrigin, bounds, strategy='best1bin', seed=42)

print("\n--- 差分进化结果 ---")
print(f"全局最小值位置: {result.x}")
print(f"函数最小值: {result.fun}")
print(f"迭代次数: {result.nit}")

# 可视化 (仅用于演示函数形状)
x = np.linspace(-5.12, 5.12, 100)
y = np.linspace(-5.12, 5.12, 100)
X, Y = np.meshgrid(x, y)
Z = 10 * 2 + (X**2 - 10 * np.cos(2 * np.pi * X)) + (Y**2 - 10 * np.cos(2 * np.pi * Y))

plt.figure(figsize=(8, 6))
plt.contourf(X, Y, Z, levels=50, cmap='viridis')
plt.colorbar(label='Objective Value')
plt.scatter(result.x[0], result.x[1], color='red', marker='*', s=200, label='Found Minimum')
plt.title("Differential Evolution on Rastrigin Function")
plt.legend()
plt.show()
```

### 💻 Python 实现示例

这里提供两个数模竞赛中**出场率极高**的通用代码模板。

#### 1. 熵权法 + TOPSIS (超级评价模型)
*这是解决“评价类”问题的标准答案。先用熵权法定权重，再用 TOPSIS 排名。*

```python
import numpy as np
import pandas as pd

def topsis_entropy(data, weight=None):
    """
    data: numpy array, 每一行是一个样本，每一列是一个指标
    注意：输入前必须先进行正向化处理（把极小型指标转化为极大型）
    """
    # 1. 归一化 (Normalization)
    # 使用向量归一化，分母为平方和的根
    norm_data = data / np.sqrt((data**2).sum(axis=0))
    
    # 2. 熵权法计算权重 (如果未指定权重)
    if weight is None:
        # 计算比重
        P = norm_data / norm_data.sum(axis=0)
        # 处理 log(0)
        epsilon = 1e-6
        # 计算熵值
        E = -1/np.log(len(data)) * (P * np.log(P + epsilon)).sum(axis=0)
        # 计算差异系数
        D = 1 - E
        # 归一化得到权重
        weight = D / D.sum()
        print("自动计算的熵权权重:", weight)
    
    # 3. 计算加权矩阵
    Z = norm_data * weight
    
    # 4. 确定正负理想解
    Z_max = Z.max(axis=0) # 理想解
    Z_min = Z.min(axis=0) # 负理想解
    
    # 5. 计算距离
    D_plus = np.sqrt(((Z - Z_max)**2).sum(axis=1))
    D_minus = np.sqrt(((Z - Z_min)**2).sum(axis=1))
    
    # 6. 计算得分 (接近 1 最好)
    score = D_minus / (D_plus + D_minus)
    
    return score, weight

# --- 使用示例 ---
# 假设有 4 个方案，3 个指标 (指标均为正向)
X = np.array([
    [100, 90, 80],
    [80, 95, 85],
    [60, 80, 90],
    [90, 70, 60]
])

scores, w = topsis_entropy(X)
print("\n各方案得分:", scores)
# 排序
ranking = np.argsort(-scores) + 1
print("排名:", ranking)
```

#### 2. 微分方程拟合 (Parametric ODE Fitting)
*这是 Problem A 的常见套路：假设物理规律已知（如 SIR 模型），但不知道参数 $\beta, \gamma$，利用真实数据反求参数。*

```python
import numpy as np
from scipy.integrate import odeint
from scipy.optimize import minimize
import matplotlib.pyplot as plt

# 1. 定义微分方程 (SIR 模型)
def sir_model(y, t, beta, gamma):
    S, I, R = y
    N = S + I + R
    dSdt = -beta * S * I / N
    dIdt = beta * S * I / N - gamma * I
    dRdt = gamma * I
    return [dSdt, dIdt, dRdt]

# 2. 准备真实数据 (比如某地 10 天的感染人数)
t_data = np.arange(0, 10, 1)
# 假设这是真实观测到的感染者人数 I
I_real = np.array([1, 3, 8, 20, 45, 90, 150, 220, 280, 310])

# 3. 定义损失函数 (预测值与真实值的误差)
def loss_func(params):
    beta, gamma = params
    # 初始值: S0=999, I0=1, R0=0
    y0 = [999, 1, 0]
    # 解方程
    solution = odeint(sir_model, y0, t_data, args=(beta, gamma))
    I_pred = solution[:, 1] # 取出 I (感染者) 的预测列
    # 返回均方误差
    return np.mean((I_real - I_pred)**2)

# 4. 优化求解参数
# 初始猜测 beta=0.5, gamma=0.1
res = minimize(loss_func, x0=[0.5, 0.1], bounds=[(0, 5), (0, 1)])
beta_opt, gamma_opt = res.x

print(f"拟合得到最优参数: beta={beta_opt:.4f}, gamma={gamma_opt:.4f}")

# 5. 绘图验证
t_future = np.linspace(0, 20, 100)
sol_final = odeint(sir_model, [999, 1, 0], t_future, args=(beta_opt, gamma_opt))

plt.scatter(t_data, I_real, color='red', label='Real Data')
plt.plot(t_future, sol_final[:, 1], label='Fitted Curve (I)')
plt.legend()
plt.show()
```
